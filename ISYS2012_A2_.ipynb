Step 1: Understand the Problem
Business Problem: As a university student living in Perth, it is easy to lose track of minor daily expenses like coffee, transport, and eating out. By the end of the month, I often don't know where my money went and whether I hit my savings target.

Solution: I am building a "CommBank-style Budget Breaker". This finance assistant will automatically process my monthly transaction data, categorize my expenses, calculate the percentage of my income spent on each category, and tell me if I have achieved my monthly savings goal. This helps me visualize my spending habits and adjust my budget for the next month.

Target User: University students (like myself) who need a quick, no-fuss way to track their living expenses and savings without manually calculating every receipt.

Step 2: Identify the Inputs and Outputs
Inputs:

csv_file_path (String): A CSV file containing my monthly transactions. The file must have the following columns: Date, Amount, Category, and Description.

monthly_income (Float): My total expected income for the month (e.g., from part-time gigs or allowance).

saving_goal (Float): The target amount of money I want to save by the end of the month.

Outputs:
A comprehensive summary (returned as a Python dictionary) containing:

category_totals (Dictionary): The total amount of money spent in each specific category (e.g., Groceries, Coffee).

category_percentages (Dictionary): The percentage of my monthly_income consumed by each category.

total_spent (Float): The absolute total of all expenses.

remaining_balance (Float): The money left over (monthly_income - total_spent).

goal_achieved (Boolean): True if the remaining_balance is greater than or equal to the saving_goal, otherwise False.
